# Task 1 - Give 2 examples of all String functions which should be meaningful.

```python
string_input2 = "games"
string_input2 = "play games"
print(f'Input string1: {string_input1} and string2: {string_input2}')
```

## Capitalize

```python
caps_string_input1 = string_input1.capitalize()
caps_string_input2 = string_input2.capitalize()
print(f'Capitalize funtion of string1: {caps_string_input1} and string2: {caps_string_input2}')
```

## Title Function

```python
title_string_input1 = string_input1.title()
title_string_input2 = string_input2.title()
print(f'Title funtion of string1: {title_string_input1} and string2: {title_string_input2}')
```

## Upper function

```python
upper_string_input1 = string_input1.upper()
upper_string_input2 = string_input2.upper()
print(f'Upper funtion of string1: {upper_string_input1} and string2: {upper_string_input2}')
```

## lower function

```python
lower_string_input1 = "Be The Leader"
lower_string_input2 = "TAJ MAHAL"
lower_string1 = lower_string_input1.lower()
lower_string2 = lower_string_input2.lower()
print(f'Lower funtion of string1: {lower_string1} and string2: {lower_string2}')
```

## Swap function

```python
string_input1 = "i aM iRON mAN"
string_input2 = "Spider MAN"
swap_string1 = string_input1.swapcase()
swap_string2 = string_input2.swapcase()
print(f'Swapcase funtion of string1: {swap_string1} and string2: {swap_string2}')
```

## casefold function

```python
string_input1 = "I am SPIDERMAN"
string_input2 = "I am IRON MAN"
casefold_string1 = string_input1.casefold()
casefold_string2 = string_input2.casefold()
print(f'Casfold funtion of string1: {casefold_string1} and string2: {casefold_string2}')
```

## strip functions

```python
string_input1 = " welcome"
string_input2 = " Hello "
print(f'input string1: |{string_input1}| and string2: |{string_input2}|')
lstrip_string1 = string_input1.lstrip()
lstrip_string2 = string_input2.lstrip()
print(f'Lstrip function of string1: |{lstrip_string1}| and string2: |{lstrip_string2}|')
rstrip_string1 = string_input1.rstrip()
rstrip_string2 = string_input2.rstrip()
print(f'Rstrip function of string1: |{rstrip_string1}| and string2: |{rstrip_string2}|')
strip_string1 = string_input1.strip()
strip_string2 = string_input2.strip()
print(f'Strip function of string1: |{strip_string1}| and string2: |{strip_string2}|')
```

## String validate function

```python
string_input1 = "Football"
string_input2 = "987654321"

isalnum_string1 = string_input1.isalnum()
isalnum_string2 = string_input2.isalnum()
print(f'Isalnum function of string1: {isalnum_string1} and string2: {isalnum_string2}')

isalpha_string1 = string_input1.isalpha()
isalpha_string2 = string_input2.isalpha()
print(f'Isalpha function of string1: {isalpha_string1} and string2: {isalpha_string2}')

isdigit_string1 = string_input1.isdigit()
isdigit_string2 = string_input2.isdigit()
print(f'Isdigit function of string1: {isdigit_string1} and string2: {isdigit_string2}')

string_input2="hello"
islower_string1 = string_input1.islower()
islower_string2 = string_input2.islower()
print(f'Islower function of string1: {islower_string1} and string2: {islower_string2}')

string_input2="HELLO"
isupper_string1 = string_input1.isupper()
isupper_string2 = string_input2.isupper()
print(f'Isupper function of string1: {isupper_string1} and string2: {isupper_string2}')

string_input2="   "
isspace_string1 = string_input1.isspace()
isspace_string2 = string_input2.isspace()
print(f'Isspace function of string1: {isspace_string1} and string2: {isspace_string2}')

string_input2="Play Station"
istitle_string1 = string_input1.istitle()
istitle_string2 = string_input2.istitle()
print(f'Istitle function of string1: {istitle_string1} and string2: {istitle_string2}')

string_input2="3Python"
isidentifier_string1 = string_input1.isidentifier()
isidentifier_string2 = string_input2.isidentifier()
print(f'Isidentifier function of string1: {isidentifier_string1} and string2: {isidentifier_string2}')

isascii_string1 = "Python123".isascii()
isascii_string2 = "Café".isascii()
print(f'Isascii function of string1: {isascii_string1} and string2: {isascii_string2}')

isdecimal_string1 = "12345".isdecimal()
isdecimal_string2 = "²345".isdecimal()
print(f'Isdecimal function of string1: {isdecimal_string1} and string2: {isdecimal_string2}')

isprintable_string1 = string_input1.isprintable()
isprintable_string2 = "Hello\nWorld".isprintable()
print(f'Isprintable function of string1: {isprintable_string1} and string2: {isprintable_string2}')
```

## Split function

```python
string_input1 = "Apple,Beetroot,Carrot"
string_input2 = "Hi\nWelcome\nStudent"
string_input3 = "one two three"

split_string1 = string_input1.split(',')
split_string3 = string_input3.split()
print(f'Split function of string1: {split_string1} and string3: {split_string3}')

rsplit_string1 = string_input1.rsplit(',', 1)
rsplit_string3 = string_input3.rsplit(' ', 1)
print(f'Rsplit function of string1: {rsplit_string1} and string3: {rsplit_string3}')

splitlines_string1 = string_input1.splitlines()
splitlines_string2 = string_input2.splitlines()
print(f'Splitlines function of string1: {splitlines_string1} and string2: {splitlines_string2}')

partition_string1 = string_input1.partition(',')
partition_string3 = string_input3.partition(' ')
print(f'Partition function of string1: {partition_string1} and string3: {partition_string3}')

rpartition_string1 = string_input1.rpartition(',')
rpartition_string3 = string_input3.rpartition('-')
print(f'Rpartition function of string1: {rpartition_string1} and string3: {rpartition_string3}')
```

## other string fucntions

```python
string_input1 = "python_3.py"
string_input2 = "hello world"

center_string1 = string_input1.center(40, '*')
center_string2 = string_input2.center(31, '-')
print(f'Center function of string1: {center_string1} and string2: {center_string2}')

count_string1 = string_input2.count('a')
count_string2 = string_input2.count('o')
print(f'Count function of string1: {count_string1} and string2: {count_string2}')

encode_string1 = string_input1.encode('utf-8')
encode_string2 = "Café".encode('utf-8')
print(f'Encode function of string1: {encode_string1} and string2: {encode_string2}')

endswith_string1 = string_input1.endswith('.py')
endswith_string2 = string_input1.endswith('.txt')
print(f'Endswith function of string1: {endswith_string1} and string2: {endswith_string2}')

expandtabs_string1 = "Py\tthon".expandtabs(4)
expandtabs_string2 = "Hello\tWorld".expandtabs(8)
print(f'Expandtabs function of string1: {expandtabs_string1} and string2: {expandtabs_string2}')

find_string1 = string_input1.find('on')
rfind_string2 = string_input1.rfind('m')
print(f'Find function of string1: {find_string1} and Rfind string2: {rfind_string2}')

format_string1 = "Hello, {}".format("Hulk")
format_string2 = "I have {} super heros with {} super powers".format(3, 5)
print(f'Format function of string1: {format_string1} and string2: {format_string2}')

map_data = {'name': 'Thor', 'age': 1225}
format_map_string1 = "Name: {name}".format_map(map_data)
format_map_string2 = "Age: {age}".format_map(map_data)
print(f'Format_map function of string1: {format_map_string1} and string2: {format_map_string2}')

index_string1 = string_input1.index('python')
rindex_string2 = string_input1.rindex('o')
print(f'Index function of string1: {index_string1} and Rindex string2: {rindex_string2}')

join_string1 = "-".join(["thor", "spidey", "hulk"])
join_string2 = ", ".join(["apple", "banana"])
print(f'Join function of string1: {join_string1} and string2: {join_string2}')

ljust_string1 = "Python".ljust(10, '*')
rjust_string2 = "Python".rjust(10, '*')
print(f'Ljust function of string1: {ljust_string1} and Rjust string2: {rjust_string2}')

table = str.maketrans('aeiou', '12345')
translate_string1 = "ironman".translate(table)
translate_string2 = "python".translate(table)
print(f'Translate function of string1: {translate_string1} and string2: {translate_string2}')

removeprefix_string1 = "unhappy".removeprefix("un")
removeprefix_string2 = "reentry".removeprefix("re")
print(f'Removeprefix function of string1: {removeprefix_string1} and string2: {removeprefix_string2}')

removesuffix_string1 = string_input1.removesuffix(".py")
removesuffix_string2 = "document.mp4".removesuffix(".mp4")
print(f'Removesuffix function of string1: {removesuffix_string1} and string2: {removesuffix_string2}')

replace_string1 = "hello world".replace("world", "python")
replace_string2 = "banana".replace("a", "o", 2)
print(f'Replace function of string1: {replace_string1} and string2: {replace_string2}')

startswith_string1 = string_input1.startswith("python")
startswith_string2 = string_input1.startswith("script")
print(f'Startswith function of string1: {startswith_string1} and string2: {startswith_string2}')

zfill_string1 = "42".zfill(5)
zfill_string2 = "-42".zfill(5)
print(f'Zfill function of string1: {zfill_string1} and string2: {zfill_string2}')
```
