## 7. user defined functions with for loop and if condition

```python
def filter_vowels_name(names):
    vowels = ['a','e','i','o','u']
    vowels_list=[]
    for letter in names:
        if letter in vowels:
            vowels_list.append(letter)
    return vowels_list

vowels_list = filter_vowels_name("pithon")
print(f"Vowels letter extracted: {vowels_list}")
```

# TASK 3 - Rock, Paper, Scissor Game

```python
import random

game = True
status = False
while game:
    options = ['rock','paper','scissor']
    user_input=input("enter your option (rock, paper,scissor):")
    if user_input.lower() not in options:
        print ("You have enter invalid options, provide the correct option")
        continue
    computer_input = random.choice(options)
    print(f"computr option is: {computer_input}")
    if computer_input == user_input:
        print(f'Both You n Computer choose same options, sipping this comparision')
    elif computer_input == 'rock' and user_input == 'paper':
        print(f'You Won the Game')
        status = True
    elif computer_input == 'paper' and user_input == 'scissor':
        print(f'You Won the Game')
        status = True
    elif computer_input == 'scissor' and user_input == 'rock':
        print(f'You Won the Game')
        status = True
    elif computer_input == 'paper' and user_input == 'rock':
        print(f'You Lose the Game')
        status = True
    elif computer_input == 'rock' and user_input == 'scissor':
        print(f'You Lose the Game')
        status = True
    elif computer_input == 'scissor' and user_input == 'paper':
        print(f'You Lose the Game')
        status = True
    else:
       status = False
        
    if status:
        game_on=input("Do you want to Continue the game:")
        if game_on.lower() == 'yes':
            game = True
        else:
            game = False
```

# Task 3 - Word Guessing Game (Mini Hangman)

```python
import random
words = ["IRONMAN","SPIDER","HULK","THOR"]
secret_word = random.choice(words).strip().lower()
word_length = "_"*len(secret_word)
guessed_letters = []
max_attempts = 6
wrong_attempts = 0

while wrong_attempts < max_attempts:
    
    display_word = []
    for letter in secret_word:
        if letter in guessed_letters:
            display_word.append(letter)
        else:
            display_word.append("_")
        
    current_state = " ".join(display_word)
    print(f"\nWord to guess: {current_state}")
    print(f"Attempts remaining: {max_attempts - wrong_attempts}")
        
    if "_" not in display_word:
        print(f"\n🎉 Congratulations! You guessed the word correctly: '{secret_word.upper()}'")
        break



    user_guess = input(f"Enter the letter to guess word:").lower()
    try:
        if len(user_guess) != 1:
            raise ValueError("Please enter exactly one letter at a time.")
        if not user_guess.isalpha():
            raise TypeError("Invalid input! Only alphabetic characters are allowed.")
        if user_guess in guessed_letters:
            raise KeyError("You have already guessed that letter. Try a different one!")
                
    except (ValueError, TypeError, KeyError) as error:
        print(f"⚠️ {error}")
        continue

    guessed_letters.append(user_guess)
    if user_guess in secret_word:
        print(f"Good job! '{user_guess}' is in the word.")
    else:
        wrong_attempts += 1
        print(f"Sorry, '{user_guess}' is not in the word.")

if wrong_attempts == max_attempts:
    print(f"\n❌ Game Over! You ran out of attempts. The word was: '{secret_word.upper()}'")
```
