## List
- 1D
- Heterogeneous
- Duplicate values
- Indexing & slicing
- Mutable

```python
empty_cart=[]
print(f'{empty_cart}')
type(empty_cart)
```

```python
items_to_buy = ["Apple","orange","banana","Apple","pomgranite","grapes",6,8.9,4]
```

```python
print(f'{items_to_buy[0]}')
print(f'{items_to_buy[-1]}')
print(f'{items_to_buy[5]}')
print(f'{items_to_buy[0:2]}')
print(f'{items_to_buy[-7:-4]}')
print(f'{items_to_buy[:4]}')
print(f'{items_to_buy[:5:2]}') # interval of 2
print(f'{items_to_buy[::-1]}') # interval -1 indicate reverse order
items_to_buy[4]="Pomogranites"
print(f'{items_to_buy}')
```

## Sets
- 1D
- Heterogeneous
- No Duplicate values
- mutable
- No Indxing & Slicing

```python
empty_cart = set()
type(empty_cart)
```

```python
items_to_buy_2 = {"Apple","orange","banana","Apple","pomgranite","grapes",6,8.9,4}
print(f'{items_to_buy_2}')
type(items_to_buy_2)
```

## Tuple
- 1D
- Heterogeneous
- Duplicate value allowed
- Indexing and slicing
- immutable

```python
empty_tuple=()
print(f'{empty_tuple}')
items_to_buy_3 = ("Apple","orange","banana","Apple","pomgranite","grapes",6,8.9,4)
print(f'{items_to_buy_3}')
type(items_to_buy_3)
print(f'{items_to_buy_3[0]}')
print(f'{items_to_buy_3[-1]}')
print(f'{items_to_buy_3[0:2]}')
items_to_buy_3[4]="Pomogranites"
```

## Dictionary
- key:value pair
- 2D
- Heterogeneous
- Mutable
- Indexing & slicing is not possible
- Don't create Duplicate key, but you can create Duplicate values

```python
empty_dict = {}
type (empty_dict)
```

```python
vegitables={"Apple":3,"Orange":2,"Onion":5}
type(vegitables)
```
