# Data Structure

## List
- 1D
- Heterogeneous
- Duplicate values
- Indexing & slicing
- Mutable

```python
empty_cart=[]
type(empty_cart)
```

```python
items_to_buy = ["Apple","orange","banana","Apple","pomgranite","grapes",6,8.9,4]
print(f'{items_to_buy[0]}')
print(f'{items_to_buy[-1]}')
print(f'{items_to_buy[5]}')
print(f'{items_to_buy[0:2]}')
print(f'{items_to_buy[-7:-4]}')
print(f'{items_to_buy[:4]}')
print(f'{items_to_buy[:5:2]}') # interval of 2
print(f'{items_to_buy[::-1]}') # interval -1 indicate reverse order
items_to_buy[4]="Pomogranites"
print(f'{items_to_buy}')
```

```python
print(f'items_to_buy before clear: {items_to_buy}')
items_to_buy.clear() # clear all the values in list
print(f'items_to_buy after clear: {items_to_buy}')
items_to_buy = ["dragon","blueberry","avacoda","blueberry",6,8.9,4]
print(f'new items_to_buy: {items_to_buy}')
new_copy=items_to_buy.copy()
new_copy.append("pear")
print(f'after copy function: {new_copy} of items_to_buy: {items_to_buy}')
print(f'count of items to buy: {items_to_buy.count("blueberry")}')
new_list=["fruits","vegetables","groceries"]
items_to_buy.extend(new_list)
print(f'items to buy extend: {items_to_buy}')
print(f'index of fruits in items to buy: {items_to_buy.index("fruits")}')
items_to_buy.insert(3, "new_fruit")
print(f'items to buy after insert: {items_to_buy}')
removd_item = items_to_buy.pop()
print(f'removed item: {removd_item}')
items_to_buy.remove("blueberry")
print(f'items to buy after remove: {items_to_buy}')
items_to_buy.reverse()
print(f'items to buy after reverse: {items_to_buy}')
new_list.sort()
print(f'new_list after sort: {new_list}')
```

## Set
- 1D
- Heterogeneous
- No Duplicate values
- mutable
- No Indxing & Slicing

```python
empty_cart = set()
type(empty_cart)
items_to_buy_2 = {"Apple","orange","banana","Apple","pomgranite","grapes",6,8.9,4}
print(f'{items_to_buy_2}')
print(f'type of item to buy 2: {type(items_to_buy_2)}')
copy_items=items_to_buy_2.copy()
copy_items.add("new_item")
print(f'after adding new value to copy item: {copy_items} and orginal items to buy2 is: {items_to_buy_2}')
new_items={"dragon","pear","Apple"}
diff=items_to_buy_2.difference(new_items)
print(f'difference: {diff} an original items to buy2: {items_to_buy_2}')
items_to_buy_2.difference_update(new_items)
print(f'difference update of the items to buy2: {items_to_buy_2}')
items_to_buy_2.discard("pear")
print(f'discard function - items to buy2: {items_to_buy_2}')
# items_to_buy_2.remove("pear") # it will fail as there is no pear value in the set
items_to_buy_2.remove("orange")
print(f'remove function - items to buy2: {items_to_buy_2}')

print(f'is disjoint of {items_to_buy_2.isdisjoint(new_items)}')
print(f'issubset of {items_to_buy_2.issubset(copy_items)}')
print(f'issuperset of {items_to_buy_2.issuperset(new_items)}')
last_value = items_to_buy_2.pop()
print(f'last value from items using pop: {last_value}')
sym_items=items_to_buy_2.symmetric_difference(new_items)
print(f'symmetric_difference of items to buy2 with new_items: {sym_items}')
items_to_buy_2.symmetric_difference_update(new_items)
print(f'symmetric_difference_upte of items to buy2 with new_items: {items_to_buy_2}')

new_intersection=items_to_buy_2.intersection(new_items)
print(f'intersection of items to buy2 an new_items: {new_intersection}')
items_to_buy_2.intersection_update(new_items)
print(f'intersection update of items to buy2 an new_items: {items_to_buy_2}')

union_items = items_to_buy_2.union({"mango","kiwi"})
print(f'union_items of items to buy2 : {union_items}')
items_to_buy_2.update({"mango","kiwi"})
print(f'update to items to buy2 : {items_to_buy_2}')
items_to_buy_2.clear()
print(f'after clear of items to buy2 : {len(items_to_buy_2)}')
```

## Tuple
- 1D
- Heterogeneous
- Duplicate value allowed
- Indexing and slicing
- immutable

```python
empty_tuple=()
print(f'{empty_tuple}')
items_to_buy_3 = ("Apple","orange","banana","Apple","pomgranite","grapes",6,8.9,4)
print(f'{items_to_buy_3}')
type(items_to_buy_3)
print(f'{items_to_buy_3[0]}')
print(f'{items_to_buy_3[-1]}')
print(f'{items_to_buy_3[0:2]}')
print(f'count of Apple: {items_to_buy_3.count("Apple")}')
print(f'index of Apple: {items_to_buy_3.index("Apple")}')
```

## Dictionary
- key:value pair
- 2D
- Heterogeneous
- Mutable
- Indexing & slicing is not possible
- Don't create Duplicate key, but you can create Duplicate values

```python
vegetables={"Apple":3,"Orange":2,"Onion":5}
print(f'values of vegitables:{vegetables.values()}')
print(f'keys of vegitables:{vegetables.keys()}')

dict1=vegetables.copy()
print(f'copy of vegitables:{dict1}')
list1=["Tomato","beans"]
list2 = [3,4]
dict2 = dict.fromkeys(list1,2)
print(f'new dictionary from list: {dict2}')
print(f'vlue of the APPLE : {vegetables.get("Apple")}')
print(f'get key an values in dictionary: {vegetables.items()}')

print(f'pop Orange in dictionary: {vegetables.pop("Orange")}, current Dictionary:{vegetables}')
print(f'pop Orange in dictionary: {vegetables.popitem()}, current Dictionary:{vegetables}')
print(f'setdefault function: {vegetables.setdefault("Apple", 10)},, current Dictionary:{vegetables}')
print(f'setdefault function: {vegetables.setdefault("Onion", 5)}, current Dictionary:{vegetables}')
new_dict={"carrot":2, "beetroot":1}
vegetables.update(new_dict)
print(f'updated Dictionary:{vegetables}')
vegetables.clear()
print(f'after clear vegetables:{vegetables}')
```
